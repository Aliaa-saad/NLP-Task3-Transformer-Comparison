```python

import os
import json
import time
import random
import numpy as np
import torch
import matplotlib.pyplot as plt
import seaborn as sns

from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup
)

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)

# Reproducibility
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("PyTorch version:", torch.__version__)
print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("WARNING: GPU is not enabled.")
    print("Go to Runtime > Change runtime type > T4 GPU")

# Model configuration
MODEL_CHECKPOINT = "distilbert-base-uncased"

NUM_LABELS = 3
MAX_LENGTH = 256

TRAIN_BATCH_SIZE = 16
EVAL_BATCH_SIZE = 32

NUM_EPOCHS = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01

USE_AMP = torch.cuda.is_available()

BEST_MODEL_DIR = MODEL_DIR / "distilbert_yelp_sentiment"

print("\nModel configuration ready!")
```

```python

# ==========================================
# LOAD PRETRAINED TOKENIZER
# ==========================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_CHECKPOINT
)

print("Tokenizer loaded successfully!")

print("Vocabulary size:", tokenizer.vocab_size)
print("Maximum sequence length:", MAX_LENGTH)

# Custom PyTorch Dataset
class YelpSentimentDataset(Dataset):

    def __init__(self, dataframe):

        self.texts = (
            dataframe["text"]
            .astype(str)
            .tolist()
        )

        self.labels = (
            dataframe["label"]
            .astype(int)
            .tolist()
        )

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, index):

        return {
            "text": self.texts[index],
            "label": self.labels[index]
        }


# Dynamic padding collator
def collate_batch(batch):

    texts = [item["text"] for item in batch]

    labels = torch.tensor(
        [item["label"] for item in batch],
        dtype=torch.long
    )

    encoded = tokenizer(
        texts,
        padding=True,
        truncation=True,
        max_length=MAX_LENGTH,
        return_tensors="pt"
    )

    encoded["labels"] = labels

    return encoded


# Create datasets
train_dataset = YelpSentimentDataset(train_df)
val_dataset = YelpSentimentDataset(val_df)
test_dataset = YelpSentimentDataset(test_df)

# Create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=TRAIN_BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_batch,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_dataset,
    batch_size=EVAL_BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_batch,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=EVAL_BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_batch,
    num_workers=0
)

print("\nDataLoaders created successfully!")

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))

# Inspect one batch
sample_batch = next(iter(train_loader))

print("\nInput IDs shape:", sample_batch["input_ids"].shape)
print("Attention Mask shape:", sample_batch["attention_mask"].shape)
print("Labels shape:", sample_batch["labels"].shape)
```

```python

def evaluate_model(model, data_loader):

    model.eval()

    all_predictions = []
    all_labels = []

    total_loss = 0

    with torch.inference_mode():

        for batch in tqdm(data_loader, desc="Evaluating"):

            batch = {
                key: value.to(device)
                for key, value in batch.items()
            }

            with torch.amp.autocast(
                device_type="cuda",
                enabled=USE_AMP
            ):

                outputs = model(**batch)

            total_loss += outputs.loss.item()

            predictions = torch.argmax(
                outputs.logits,
                dim=1
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                batch["labels"].cpu().numpy()
            )

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        )
    )

    return {
        "loss": total_loss / len(data_loader),
        "accuracy": accuracy,
        "precision_macro": precision,
        "recall_macro": recall,
        "f1_macro": f1,
        "predictions": np.array(all_predictions),
        "labels": np.array(all_labels)
    }
```

```python

# ==========================================
# CLEAN TRAINING INITIALIZATION
# ==========================================

import gc
import random
import numpy as np
import torch
import shutil

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup
)

# 1. Reset random seeds
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

assert torch.cuda.is_available(), "GPU is required!"

device = torch.device("cuda")

# 2. Recreate reproducible training DataLoader
generator = torch.Generator()
generator.manual_seed(SEED)

train_loader = DataLoader(
    train_dataset,
    batch_size=TRAIN_BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_batch,
    num_workers=0,
    pin_memory=True,
    generator=generator
)

# 3. Load ORIGINAL pretrained weights
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_CHECKPOINT,
    num_labels=NUM_LABELS,
    label2id=LABEL2ID,
    id2label=ID2LABEL
).to(device)

# 4. Fresh optimizer
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

# 5. Fresh scheduler
total_training_steps = len(train_loader) * NUM_EPOCHS

warmup_steps = int(0.1 * total_training_steps)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_training_steps
)

# 6. Fresh AMP scaler
scaler = torch.amp.GradScaler(
    "cuda",
    enabled=USE_AMP
)

# 7. Use a new checkpoint directory
BEST_MODEL_DIR = (
    MODEL_DIR / "distilbert_yelp_sentiment_clean"
)

if BEST_MODEL_DIR.exists():
    shutil.rmtree(BEST_MODEL_DIR)

BEST_MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# 8. Reset training tracking
training_history = []
best_val_f1 = -1

print("Clean training initialization completed!")
print("GPU:", torch.cuda.get_device_name(0))
print("Training batches:", len(train_loader))
print("Total steps:", total_training_steps)
print("Warmup steps:", warmup_steps)
print("Initial learning rate:", optimizer.param_groups[0]["lr"])
print("Checkpoint directory:", BEST_MODEL_DIR)
```

```python

# ==========================================
# CLEAN FINE-TUNING TRAINING LOOP
# ==========================================

from tqdm.auto import tqdm
import time

training_start_time = time.perf_counter()

for epoch in range(NUM_EPOCHS):

    print(f"\n{'='*50}")
    print(f"Epoch {epoch + 1}/{NUM_EPOCHS}")
    print(f"{'='*50}")

    model.train()

    total_train_loss = 0

    progress_bar = tqdm(
        train_loader,
        desc=f"Training Epoch {epoch + 1}"
    )

    for batch in progress_bar:

        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast(
            device_type="cuda",
            enabled=USE_AMP
        ):

            outputs = model(**batch)
            loss = outputs.loss

        scaler.scale(loss).backward()

        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        scaler.step(optimizer)
        scaler.update()

        scheduler.step()

        total_train_loss += loss.item()

        progress_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            lr=f"{scheduler.get_last_lr()[0]:.2e}"
        )

    avg_train_loss = (
        total_train_loss / len(train_loader)
    )

    # Validation
    val_metrics = evaluate_model(
        model,
        val_loader
    )

    print(f"\nTraining Loss: {avg_train_loss:.4f}")
    print(f"Validation Loss: {val_metrics['loss']:.4f}")
    print(f"Validation Accuracy: {val_metrics['accuracy']:.4f}")
    print(f"Validation Macro-F1: {val_metrics['f1_macro']:.4f}")

    training_history.append({
        "epoch": epoch + 1,
        "train_loss": avg_train_loss,
        "validation_loss": val_metrics["loss"],
        "validation_accuracy": val_metrics["accuracy"],
        "validation_precision_macro": val_metrics["precision_macro"],
        "validation_recall_macro": val_metrics["recall_macro"],
        "validation_f1_macro": val_metrics["f1_macro"]
    })

    if val_metrics["f1_macro"] > best_val_f1:

        best_val_f1 = val_metrics["f1_macro"]

        model.save_pretrained(BEST_MODEL_DIR)
        best_tokenizer = tokenizer

        tokenizer.save_pretrained(BEST_MODEL_DIR)

        print("Best model saved successfully!")

    gc.collect()
    torch.cuda.empty_cache()

training_time_seconds = (
    time.perf_counter() - training_start_time
)

training_history_df = pd.DataFrame(training_history)

print("\nTraining completed!")

display(training_history_df)

print(
    f"Training time: "
    f"{training_time_seconds / 60:.2f} minutes"
)

print("Best Validation Macro-F1:", best_val_f1)
```

```python

# ==========================================
# LOAD BEST MODEL
# ==========================================

best_model = AutoModelForSequenceClassification.from_pretrained(
    BEST_MODEL_DIR
).to(device)

best_tokenizer = AutoTokenizer.from_pretrained(
    BEST_MODEL_DIR
)

best_model.eval()

print("Best model loaded successfully!")

# Final test evaluation
test_metrics = evaluate_model(
    best_model,
    test_loader
)

print("\nFINAL TEST RESULTS")

for metric in [
    "accuracy",
    "precision_macro",
    "recall_macro",
    "f1_macro"
]:

    print(
        f"{metric}: {test_metrics[metric]:.4f}"
    )
```

```python

# ==========================================
# CLASSIFICATION REPORT - FIXED
# ==========================================

import pandas as pd
from sklearn.metrics import classification_report

# Get actual test predictions
y_true = test_metrics["labels"]
y_pred = test_metrics["predictions"]

print("Number of test samples:", len(y_true))
print("Generating classification report...")

report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1, 2],
    target_names=[
        "Negative",
        "Neutral",
        "Positive"
    ],
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

print("\nClassification Report:")
display(report_df)

print("\nClassification report completed!")
```

```python

# ==========================================
# INFERENCE LATENCY
# ==========================================

best_model.eval()

latency_texts = test_df["text"].head(55).tolist()

latencies = []

# Warm-up
for text in latency_texts[:5]:

    inputs = best_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LENGTH
    ).to(device)

    with torch.inference_mode():

        _ = best_model(**inputs)

if device.type == "cuda":
    torch.cuda.synchronize()

# Actual measurements
for text in latency_texts[5:]:

    if device.type == "cuda":
        torch.cuda.synchronize()

    start = time.perf_counter()

    inputs = best_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LENGTH
    ).to(device)

    with torch.inference_mode():

        _ = best_model(**inputs)

    if device.type == "cuda":
        torch.cuda.synchronize()

    elapsed = time.perf_counter() - start

    latencies.append(elapsed * 1000)

average_latency_ms = float(np.mean(latencies))
median_latency_ms = float(np.median(latencies))

# Parameter count
total_parameters = sum(
    p.numel()
    for p in best_model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in best_model.parameters()
    if p.requires_grad
)

print("INFERENCE PERFORMANCE")
print(f"Average latency: {average_latency_ms:.2f} ms")
print(f"Median latency: {median_latency_ms:.2f} ms")
print(f"Total parameters: {total_parameters:,}")
print(f"Trainable parameters: {trainable_parameters:,}")
print(f"Device: {device}")
```

```python

# ==========================================
# INFERENCE FUNCTION FOR STREAMLIT
# ==========================================

def predict_sentiment(text):

    if not isinstance(text, str) or not text.strip():
        raise ValueError("Please enter a valid review.")

    best_model.eval()

    inputs = best_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LENGTH
    ).to(device)

    with torch.inference_mode():

        outputs = best_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )[0]

    predicted_id = int(
        torch.argmax(probabilities).item()
    )

    sentiment = ID2LABEL[predicted_id]

    confidence = float(
        probabilities[predicted_id].item()
    )

    return {
        "model": "DistilBERT",
        "sentiment": sentiment,
        "confidence": round(confidence * 100, 2),
        "probabilities": {
            ID2LABEL[i]: round(
                float(probabilities[i].item()) * 100,
                2
            )
            for i in range(NUM_LABELS)
        }
    }


# Test prediction
sample_review = (
    "The food was amazing and the staff were very friendly!"
)

prediction = predict_sentiment(sample_review)

print("Input:", sample_review)
print("\nPrediction:")

for key, value in prediction.items():
    print(f"{key}: {value}")
```
